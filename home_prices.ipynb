```python
import pandas as pd 
import numpy as np 
from sklearn import linear_model
import math
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("homeprice(2).csv")
df
```

```python
# now we can see we have one missimg value , so the problem is we have to find the messing value ,so we are using median function.
df.bedrooms.median()
```

```python
#its an float value we have to change in integer , so we use math function.
median_bedrooms = math.floor(df.bedrooms.median())
median_bedrooms
```

```python
# we had fillna  value NA space through new value we found .
df.bedrooms =df.bedrooms.fillna(median_bedrooms)
```

```python
# as we can see complited data set
df
```

```python
#now we gonna do linear_model to train our data. 
reg = linear_model.LinearRegression()
reg.fit(df[["area","bedrooms","age"]],df.price)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.predict([[3000,3,15]])
```
